````python
# ===== CELL 3: Utility Functions =====

import warnings
warnings.filterwarnings('ignore')

import logging
import sys
from datetime import datetime
import psutil
import gc

class ColoredFormatter(logging.Formatter):
    """Formatter ملون للوقات"""

    COLORS = {
        'DEBUG': '\033[94m',    # Blue
        'INFO': '\033[92m',     # Green
        'WARNING': '\033[93m',  # Yellow
        'ERROR': '\033[91m',    # Red
        'CRITICAL': '\033[95m', # Magenta
        'RESET': '\033[0m'
    }

    def format(self, record):
        log_color = self.COLORS.get(record.levelname, self.COLORS['RESET'])
        record.levelname = f"{log_color}{record.levelname}{self.COLORS['RESET']}"
        return super().format(record)

def setup_logger(name='FakeNewsDetector', log_file=None):
    """إعداد logger محسّن"""
    logger = logging.getLogger(name)
    logger.setLevel(logging.DEBUG)

    # Console handler
    console_handler = logging.StreamHandler(sys.stdout)
    console_handler.setLevel(logging.INFO)
    console_formatter = ColoredFormatter(
        '%(asctime)s | %(levelname)s | %(message)s',
        datefmt='%H:%M:%S'
    )
    console_handler.setFormatter(console_formatter)
    logger.addHandler(console_handler)

    # File handler
    if log_file:
        file_handler = logging.FileHandler(log_file)
        file_handler.setLevel(logging.DEBUG)
        file_formatter = logging.Formatter(
            '%(asctime)s | %(levelname)s | %(name)s | %(message)s'
        )
        file_handler.setFormatter(file_formatter)
        logger.addHandler(file_handler)

    return logger

# إنشاء logger عام
log_file = os.path.join(BASE_DIR, 'logs', 'training', f'session_{datetime.now():%Y%m%d_%H%M%S}.log')
logger = setup_logger(log_file=log_file)

logger.info("Logger initialized successfully")

# دالة لتنظيف الذاكرة
def clear_memory():
    """تنظيف ذاكرة GPU و RAM"""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

# دالة لمراقبة الموارد
def print_memory_usage():
    """طباعة استهلاك الذاكرة"""
    ram = psutil.virtual_memory()
    logger.info(f"RAM Usage: {ram.percent}% ({ram.used/1e9:.2f}/{ram.total/1e9:.2f} GB)")

    if torch.cuda.is_available():
        gpu_mem = torch.cuda.memory_allocated() / 1e9
        gpu_max = torch.cuda.max_memory_allocated() / 1e9
        logger.info(f"GPU Memory: {gpu_mem:.2f} GB (Peak: {gpu_max:.2f} GB)")

# دالة لحفظ Checkpoint
def save_checkpoint(model, optimizer, epoch, metrics, filepath):
    """حفظ checkpoint كامل"""
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'metrics': metrics,
        'timestamp': datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    }
    torch.save(checkpoint, filepath)
    logger.info(f"Checkpoint saved: {filepath}")

# دالة للتحميل
def load_checkpoint(filepath, model, optimizer=None):
    """تحميل checkpoint"""
    checkpoint = torch.load(filepath, map_location='cpu')
    model.load_state_dict(checkpoint['model_state_dict'])

    if optimizer:
        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])

    logger.info(f"Checkpoint loaded from epoch {checkpoint['epoch']}")
    return checkpoint['epoch'], checkpoint['metrics']

# دالة لعرض تقدم التدريب
from tqdm.auto import tqdm

class TrainingProgress:
    """فئة لتتبع تقدم التدريب"""

    def __init__(self, total_epochs):
        self.total_epochs = total_epochs
        self.history = {
            'train_loss': [],
            'train_acc': [],
            'val_loss': [],
            'val_acc': [],
            'learning_rate': []
        }
        self.best_val_acc = 0.0
        self.best_epoch = 0

    def update(self, epoch, train_loss, train_acc, val_loss, val_acc, lr):
        """تحديث السجل"""
        self.history['train_loss'].append(train_loss)
        self.history['train_acc'].append(train_acc)
        self.history['val_loss'].append(val_loss)
        self.history['val_acc'].append(val_acc)
        self.history['learning_rate'].append(lr)

        if val_acc > self.best_val_acc:
            self.best_val_acc = val_acc
            self.best_epoch = epoch
            return True  # إشارة لحفظ النموذج
        return False

    def plot(self, save_path=None):
        """رسم تقدم التدريب"""
        import matplotlib.pyplot as plt

        fig, axes = plt.subplots(2, 2, figsize=(15, 10))

        # Loss
        axes[0, 0].plot(self.history['train_loss'], label='Train Loss', marker='o')
        axes[0, 0].plot(self.history['val_loss'], label='Val Loss', marker='s')
        axes[0, 0].set_title('Loss over Epochs', fontsize=14, fontweight='bold')
        axes[0, 0].set_xlabel('Epoch')
        axes[0, 0].set_ylabel('Loss')
        axes[0, 0].legend()
        axes[0, 0].grid(alpha=0.3)

        # Accuracy
        axes[0, 1].plot([x*100 for x in self.history['train_acc']], label='Train Acc', marker='o')
        axes[0, 1].plot([x*100 for x in self.history['val_acc']], label='Val Acc', marker='s')
        axes[0, 1].set_title('Accuracy over Epochs', fontsize=14, fontweight='bold')
        axes[0, 1].set_xlabel('Epoch')
        axes[0, 1].set_ylabel('Accuracy (%)')
        axes[0, 1].legend()
        axes[0, 1].grid(alpha=0.3)

        # Learning Rate
        axes[1, 0].plot(self.history['learning_rate'], marker='o', color='purple')
        axes[1, 0].set_title('Learning Rate Schedule', fontsize=14, fontweight='bold')
        axes[1, 0].set_xlabel('Epoch')
        axes[1, 0].set_ylabel('Learning Rate')
        axes[1, 0].set_yscale('log')
        axes[1, 0].grid(alpha=0.3)

        # Summary
        axes[1, 1].axis('off')
        summary_text = f"""
        Training Summary
        ════════════════════════════

        Total Epochs: {len(self.history['train_loss'])}

        Best Validation Accuracy:
        {self.best_val_acc*100:.2f}% (Epoch {self.best_epoch})

        Final Train Accuracy:
        {self.history['train_acc'][-1]*100:.2f}%

        Final Validation Accuracy:
        {self.history['val_acc'][-1]*100:.2f}%

        Final Train Loss: {self.history['train_loss'][-1]:.4f}
        Final Val Loss: {self.history['val_loss'][-1]:.4f}
        """
        axes[1, 1].text(0.1, 0.5, summary_text, fontsize=11, family='monospace',
                       verticalalignment='center')

        plt.tight_layout()

        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            logger.info(f"Training plot saved: {save_path}")

        plt.show()

logger.info("✅ Utility functions loaded")
print_memory_usage()
````

````python
# ===== CELL 7: Text Stream Architecture =====
import torch
import torch.nn as nn
from transformers import BertModel, BertTokenizer, BertConfig

class TextStream(nn.Module):
    """
    Text Stream: BERT + BiLSTM + Self-Attention
    للمعالجة العميقة للنصوص
    """

    def __init__(
        self,
        bert_model_name='bert-base-uncased',
        hidden_dim=256,
        output_dim=768,
        dropout=0.3,
        freeze_bert_layers=8
    ):
        super(TextStream, self).__init__()

        logger.info(f"🔨 Building Text Stream with {bert_model_name}...")

        # تحميل BERT
        self.bert = BertModel.from_pretrained(bert_model_name)

        # تجميد الطبقات الأولى للسرعة
        if freeze_bert_layers > 0:
            for param in list(self.bert.encoder.layer[:freeze_bert_layers].parameters()):
                param.requires_grad = False
            logger.info(f"   ❄️  Froze first {freeze_bert_layers} BERT layers")

        # BiLSTM لالتقاط السياق الزمني
        self.bilstm = nn.LSTM(
            input_size=768,  # BERT hidden size
            hidden_size=hidden_dim,
            num_layers=2,
            batch_first=True,
            bidirectional=True,
            dropout=dropout
        )

        # Self-Attention Mechanism
        self.attention_weight = nn.Linear(hidden_dim * 2, 1)

        # Output layers
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * 2, output_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )

        self.layer_norm = nn.LayerNorm(output_dim)

    def attention(self, lstm_output):
        """
        Self-Attention mechanism

        Args:
            lstm_output: [batch, seq_len, hidden*2]

        Returns:
            context: [batch, hidden*2]
        """
        # حساب أوزان الانتباه
        attention_scores = self.attention_weight(lstm_output)  # [batch, seq_len, 1]
        attention_weights = torch.softmax(attention_scores, dim=1)  # [batch, seq_len, 1]

        # Weighted sum
        context_vector = torch.sum(attention_weights * lstm_output, dim=1)  # [batch, hidden*2]

        return context_vector, attention_weights

    def forward(self, input_ids, attention_mask):
        """
        Forward pass

        Args:
            input_ids: [batch, seq_len]
            attention_mask: [batch, seq_len]

        Returns:
            output: [batch, output_dim]
        """
        # BERT encoding
        bert_output = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        # استخدام last_hidden_state
        sequence_output = bert_output.last_hidden_state  # [batch, seq_len, 768]

        # BiLSTM
        lstm_output, _ = self.bilstm(sequence_output)  # [batch, seq_len, hidden*2]

        # Self-Attention
        context_vector, attention_weights = self.attention(lstm_output)

        # Final projection
        output = self.fc(context_vector)  # [batch, output_dim]
        output = self.layer_norm(output)

        return output

# إنشاء النموذج واختباره
logger.info("\n" + "="*60)
logger.info("Testing Text Stream Model...")
logger.info("="*60)

text_model = TextStream(
    bert_model_name='bert-base-uncased',
    hidden_dim=256,
    output_dim=768,
    dropout=0.3,
    freeze_bert_layers=8
)

# حساب عدد المعاملات
total_params = sum(p.numel() for p in text_model.parameters())
trainable_params = sum(p.numel() for p in text_model.parameters() if p.requires_grad)

logger.info(f"\n📊 Model Statistics:")
logger.info(f"   Total parameters: {total_params:,}")
logger.info(f"   Trainable parameters: {trainable_params:,}")
logger.info(f"   Frozen parameters: {total_params - trainable_params:,}")
logger.info(f"   Model size: ~{total_params * 4 / 1e6:.1f} MB")

# اختبار مع dummy input
logger.info("\n🧪 Testing with dummy input...")
dummy_input_ids = torch.randint(0, 30522, (2, 128))  # batch=2, seq_len=128
dummy_attention_mask = torch.ones(2, 128)

with torch.no_grad():
    output = text_model(dummy_input_ids, dummy_attention_mask)
    logger.info(f"   Input shape: {dummy_input_ids.shape}")
    logger.info(f"   Output shape: {output.shape}")
    logger.info(f"   ✅ Text Stream working correctly!")

# حفظ النموذج
model_path = os.path.join(PATHS['models/text_stream'], 'text_stream_init.pt')
torch.save(text_model.state_dict(), model_path)
logger.info(f"\n💾 Model saved to: {model_path}")

clear_memory()
````

````python
# ===== CELL 8: Image Stream Architecture =====
import torch
import torch.nn as nn
import torchvision.models as models
from torchvision.models import efficientnet_b3, EfficientNet_B3_Weights

class ImageStream(nn.Module):
    """
    Image Stream: EfficientNet-B3 للخصائص المكانية
    مع إمكانية إضافة Vision Transformer
    """

    def __init__(
        self,
        output_dim=512,
        dropout=0.3,
        freeze_layers=50
    ):
        super(ImageStream, self).__init__()

        logger.info("🔨 Building Image Stream with EfficientNet-B3...")

        # تحميل EfficientNet-B3 المُدرب مسبقاً
        weights = EfficientNet_B3_Weights.IMAGENET1K_V1
        self.efficientnet = efficientnet_b3(weights=weights)

        # الحصول على عدد features من آخر طبقة
        num_features = self.efficientnet.classifier[1].in_features

        # إزالة classifier الأصلي
        self.efficientnet.classifier = nn.Identity()

        # تجميد الطبقات الأولى
        if freeze_layers > 0:
            layers_to_freeze = list(self.efficientnet.features[:5].parameters())
            for param in layers_to_freeze:
                param.requires_grad = False
            logger.info(f"   ❄️  Froze first {freeze_layers} layers")

        # طبقات إضافية للمعالجة
        self.feature_processor = nn.Sequential(
            nn.Linear(num_features, 1024),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(1024, output_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )

        self.layer_norm = nn.LayerNorm(output_dim)

    def forward(self, images):
        """
        Forward pass

        Args:
            images: [batch, 3, 224, 224]

        Returns:
            output: [batch, output_dim]
        """
        # Extract features from EfficientNet
        features = self.efficientnet(images)  # [batch, num_features]

        # Process features
        output = self.feature_processor(features)  # [batch, output_dim]
        output = self.layer_norm(output)

        return output

# إنشاء النموذج واختباره
logger.info("\n" + "="*60)
logger.info("Testing Image Stream Model...")
logger.info("="*60)

image_model = ImageStream(
    output_dim=512,
    dropout=0.3,
    freeze_layers=50
)

# إحصائيات
total_params = sum(p.numel() for p in image_model.parameters())
trainable_params = sum(p.numel() for p in image_model.parameters() if p.requires_grad)

logger.info(f"\n📊 Model Statistics:")
logger.info(f"   Total parameters: {total_params:,}")
logger.info(f"   Trainable parameters: {trainable_params:,}")
logger.info(f"   Model size: ~{total_params * 4 / 1e6:.1f} MB")

# اختبار
logger.info("\n🧪 Testing with dummy input...")
dummy_images = torch.randn(2, 3, 224, 224)  # batch=2

with torch.no_grad():
    output = image_model(dummy_images)
    logger.info(f"   Input shape: {dummy_images.shape}")
    logger.info(f"   Output shape: {output.shape}")
    logger.info(f"   ✅ Image Stream working correctly!")

# حفظ
model_path = os.path.join(PATHS['models/image_stream'], 'image_stream_init.pt')
torch.save(image_model.state_dict(), model_path)
logger.info(f"\n💾 Model saved to: {model_path}")

clear_memory()
````

````python
# ===== CELL 9: Graph Stream Architecture =====
import torch
import torch.nn as nn
import torch.nn.functional as F

class GraphStream(nn.Module):
    """
    Graph Stream: معالجة بسيطة لخصائص الانتشار
    (في المشروع الكامل، يمكن استخدام GCN حقيقي)
    """

    def __init__(
        self,
        input_dim=256,
        hidden_dim=128,
        output_dim=256,
        dropout=0.3
    ):
        super(GraphStream, self).__init__()

        logger.info("🔨 Building Graph Stream...")

        # Multi-layer perceptron لمعالجة graph features
        self.mlp = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, output_dim),
            nn.ReLU()
        )

        self.layer_norm = nn.LayerNorm(output_dim)

    def forward(self, graph_features):
        """
        Forward pass

        Args:
            graph_features: [batch, input_dim]

        Returns:
            output: [batch, output_dim]
        """
        output = self.mlp(graph_features)
        output = self.layer_norm(output)

        return output

# إنشاء النموذج
logger.info("\n" + "="*60)
logger.info("Testing Graph Stream Model...")
logger.info("="*60)

graph_model = GraphStream(
    input_dim=256,
    hidden_dim=128,
    output_dim=256,
    dropout=0.3
)

# إحصائيات
total_params = sum(p.numel() for p in graph_model.parameters())

logger.info(f"\n📊 Model Statistics:")
logger.info(f"   Total parameters: {total_params:,}")
logger.info(f"   Model size: ~{total_params * 4 / 1e6:.2f} MB")

# اختبار
logger.info("\n🧪 Testing with dummy input...")
dummy_graph_feat = torch.randn(2, 256)

with torch.no_grad():
    output = graph_model(dummy_graph_feat)
    logger.info(f"   Input shape: {dummy_graph_feat.shape}")
    logger.info(f"   Output shape: {output.shape}")
    logger.info(f"   ✅ Graph Stream working correctly!")

# حفظ
model_path = os.path.join(PATHS['models/graph_stream'], 'graph_stream_init.pt')
torch.save(graph_model.state_dict(), model_path)
logger.info(f"\n💾 Model saved to: {model_path}")

clear_memory()
````

````python
# ===== CELL 10: Complete Multi-Modal Model =====
import torch
import torch.nn as nn

class MultiModalFakeNewsDetector(nn.Module):
    """
    النموذج الكامل: دمج Text + Image + Graph Streams
    """

    def __init__(
        self,
        text_output_dim=768,
        image_output_dim=512,
        graph_output_dim=256,
        fusion_hidden_dim=1024,
        num_classes=2,
        dropout=0.4
    ):
        super(MultiModalFakeNewsDetector, self).__init__()

        logger.info("🏗️  Building Complete Multi-Modal Model...")

        # المسارات الثلاثة
        self.text_stream = TextStream(
            bert_model_name='bert-base-uncased',
            hidden_dim=256,
            output_dim=text_output_dim,
            dropout=0.3,
            freeze_bert_layers=8
        )

        self.image_stream = ImageStream(
            output_dim=image_output_dim,
            dropout=0.3,
            freeze_layers=50
        )

        self.graph_stream = GraphStream(
            input_dim=256,
            hidden_dim=128,
            output_dim=graph_output_dim,
            dropout=0.3
        )

        # Cross-Modal Attention (مبسطة)
        total_dim = text_output_dim + image_output_dim + graph_output_dim

        self.attention = nn.Sequential(
            nn.Linear(total_dim, fusion_hidden_dim),
            nn.Tanh(),
            nn.Linear(fusion_hidden_dim, total_dim),
            nn.Softmax(dim=1)
        )

        # Fusion layers
        self.fusion = nn.Sequential(
            nn.Linear(total_dim, fusion_hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(fusion_hidden_dim, 512),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(dropout / 2)
        )

        # Classification head
        self.classifier = nn.Linear(256, num_classes)

        logger.info("   ✅ All streams integrated")

    def forward(self, input_ids, attention_mask, images, graph_features):
        """
        Forward pass

        Args:
            input_ids: [batch, seq_len]
            attention_mask: [batch, seq_len]
            images: [batch, 3, 224, 224]
            graph_features: [batch, 256]

        Returns:
            logits: [batch, num_classes]
        """
        # Extract features من كل stream
        text_feat = self.text_stream(input_ids, attention_mask)  # [batch, 768]
        image_feat = self.image_stream(images)  # [batch, 512]
        graph_feat = self.graph_stream(graph_features)  # [batch, 256]

        # Concatenate all features
        combined = torch.cat([text_feat, image_feat, graph_feat], dim=1)  # [batch, 1536]

        # Apply attention
        attention_weights = self.attention(combined)
        attended_features = combined * attention_weights

        # Fusion
        fused = self.fusion(attended_features)  # [batch, 256]

        # Classification
        logits = self.classifier(fused)  # [batch, num_classes]

        return logits

    def get_attention_weights(self, input_ids, attention_mask, images, graph_features):
        """الحصول على أوزان الانتباه للتفسير"""
        with torch.no_grad():
            text_feat = self.text_stream(input_ids, attention_mask)
            image_feat = self.image_stream(images)
            graph_feat = self.graph_stream(graph_features)

            combined = torch.cat([text_feat, image_feat, graph_feat], dim=1)
            attention_weights = self.attention(combined)

        return attention_weights

# إنشاء النموذج الكامل
logger.info("\n" + "="*60)
logger.info("🚀 Creating Complete Multi-Modal Model")
logger.info("="*60)

model = MultiModalFakeNewsDetector(
    text_output_dim=768,
    image_output_dim=512,
    graph_output_dim=256,
    fusion_hidden_dim=1024,
    num_classes=2,
    dropout=0.4
)

# إحصائيات شاملة
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

logger.info(f"\n📊 Complete Model Statistics:")
logger.info(f"{'='*60}")
logger.info(f"Total Parameters:     {total_params:,}")
logger.info(f"Trainable Parameters: {trainable_params:,}")
logger.info(f"Frozen Parameters:    {total_params - trainable_params:,}")
logger.info(f"Model Size:           ~{total_params * 4 / 1e6:.1f} MB (FP32)")
logger.info(f"                      ~{total_params * 2 / 1e6:.1f} MB (FP16)")
logger.info(f"{'='*60}")

# اختبار النموذج الكامل
logger.info("\n🧪 Testing Complete Model...")

# إنشاء dummy inputs
batch_size = 2
dummy_input_ids = torch.randint(0, 30522, (batch_size, 128))
dummy_attention_mask = torch.ones(batch_size, 128)
dummy_images = torch.randn(batch_size, 3, 224, 224)
dummy_graph_features = torch.randn(batch_size, 256)

with torch.no_grad():
    logits = model(
        input_ids=dummy_input_ids,
        attention_mask=dummy_attention_mask,
        images=dummy_images,
        graph_features=dummy_graph_features
    )

    probabilities = torch.softmax(logits, dim=1)
    predictions = torch.argmax(probabilities, dim=1)

logger.info(f"   Input batch size: {batch_size}")
logger.info(f"   Output logits shape: {logits.shape}")
logger.info(f"   Probabilities:\n{probabilities}")
logger.info(f"   Predictions: {predictions}")
logger.info(f"   ✅ Complete model working correctly!")

# حفظ النموذج
model_path = os.path.join(PATHS['models/fusion'], 'multimodal_model_init.pt')
torch.save(model.state_dict(), model_path)
logger.info(f"\n💾 Complete model saved to: {model_path}")

print_memory_usage()
clear_memory()
````

````python
# ===== CELL 11: Custom DataLoader =====
from torch.utils.data import Dataset, DataLoader
from transformers import BertTokenizer
from PIL import Image
import torchvision.transforms as transforms
import numpy as np
import ast

class FakeNewsDataset(Dataset):
    """
    Dataset مخصص للأخبار المزيفة متعددة الوسائط
    """

    def __init__(
        self,
        dataframe,
        tokenizer,
        max_length=256,
        img_size=224,
        transform=None
    ):
        self.df = dataframe.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.img_size = img_size

        # Image transforms
        if transform is None:
            self.transform = transforms.Compose([
                transforms.Resize((img_size, img_size)),
                transforms.ToTensor(),
                transforms.Normalize(
                    mean=[0.485, 0.456, 0.406],
                    std=[0.229, 0.224, 0.225]
                )
            ])
        else:
            self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        # ===== TEXT PROCESSING =====
        text = str(row['full_text'])

        # Tokenize
        encoding = self.tokenizer(
            text,
            max_length=self.max_length,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )

        input_ids = encoding['input_ids'].squeeze(0)
        attention_mask = encoding['attention_mask'].squeeze(0)

        # ===== IMAGE PROCESSING =====
        # في dataset الحقيقي، نحمل الصورة من المسار
        # هنا سنستخدم صورة عشوائية للتوضيح
        image = torch.randn(3, self.img_size, self.img_size)  # Dummy image

        # في المشروع الحقيقي:
        # try:
        #     img_path = row['local_image_path']
        #     image = Image.open(img_path).convert('RGB')
        #     image = self.transform(image)
        # except:
        #     image = torch.zeros(3, self.img_size, self.img_size)

        # ===== GRAPH FEATURES =====
        # إنشاء graph features بسيطة
        graph_features = torch.zeros(256)

        # ملء أول 11 عنصر ببيانات حقيقية
        graph_features[0] = row.get('num_tweets', 0) / 1000  # normalized
        graph_features[1] = np.log1p(row.get('num_tweets', 0))  # log-scaled
        graph_features[2] = 1 if row['source'] == 'source_a' else 0
        graph_features[3] = 1 if row['source'] == 'source_b' else 0
        graph_features[4] = min(row['text_length'] / 5000, 1.0)
        graph_features[5] = min(row['word_count'] / 1000, 1.0)
        graph_features[6] = row.get('num_images', 0) / 10
        graph_features[7] = row['word_count'] / max(row['full_text'].count('.'), 1)
        graph_features[8] = row['text_length'] / max(row['word_count'], 1)
        graph_features[9] = len(row['title']) / 100
        graph_features[10] = row['title'].count('!') + row['title'].count('?')

        # Label
        label = torch.tensor(row['label'], dtype=torch.long)

        return {
            'input_ids': input_ids,
            'attention_mask': attention_mask,
            'image': image,
            'graph_features': graph_features,
            'label': label,
            'article_id': row['id']
        }

# إنشاء Datasets
logger.info("📦 Creating Data Loaders...")

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# تحميل البيانات
train_df = pd.read_csv(os.path.join(PATHS['datasets/splits'], 'train.csv'))
val_df = pd.read_csv(os.path.join(PATHS['datasets/splits'], 'val.csv'))
test_df = pd.read_csv(os.path.join(PATHS['datasets/splits'], 'test.csv'))

# إنشاء Datasets
train_dataset = FakeNewsDataset(train_df, tokenizer, max_length=256)
val_dataset = FakeNewsDataset(val_df, tokenizer, max_length=256)
test_dataset = FakeNewsDataset(test_df, tokenizer, max_length=256)

# إنشاء DataLoaders
BATCH_SIZE = 16

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

logger.info(f"✅ Training batches: {len(train_loader)}")
logger.info(f"✅ Validation batches: {len(val_loader)}")
logger.info(f"✅ Test batches: {len(test_loader)}")

# اختبار DataLoader
logger.info("\n🧪 Testing DataLoader...")
sample_batch = next(iter(train_loader))

logger.info(f"   input_ids shape: {sample_batch['input_ids'].shape}")
logger.info(f"   attention_mask shape: {sample_batch['attention_mask'].shape}")
logger.info(f"   image shape: {sample_batch['image'].shape}")
logger.info(f"   graph_features shape: {sample_batch['graph_features'].shape}")
logger.info(f"   label shape: {sample_batch['label'].shape}")
logger.info(f"   ✅ DataLoader working correctly!")
````

````python
# ===== CELL 12: Training & Evaluation Functions =====
from tqdm.auto import tqdm
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix

def train_one_epoch(model, dataloader, criterion, optimizer, device, epoch):
    """
    تدريب epoch واحد
    """
    model.train()

    total_loss = 0
    all_predictions = []
    all_labels = []

    progress_bar = tqdm(dataloader, desc=f'Epoch {epoch} [Train]')

    for batch_idx, batch in enumerate(progress_bar):
        # نقل البيانات للـ device
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        images = batch['image'].to(device)
        graph_features = batch['graph_features'].to(device)
        labels = batch['label'].to(device)

        # Forward pass
        optimizer.zero_grad()
        logits = model(input_ids, attention_mask, images, graph_features)

        # حساب الخسارة
        loss = criterion(logits, labels)

        # Backward pass
        loss.backward()

        # Gradient clipping
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

        optimizer.step()

        # حساب المقاييس
        total_loss += loss.item()

        predictions = torch.argmax(logits, dim=1).cpu().numpy()
        all_predictions.extend(predictions)
        all_labels.extend(labels.cpu().numpy())

        # تحديث progress bar
        current_acc = accuracy_score(all_labels, all_predictions)
        progress_bar.set_postfix({
            'loss': f'{loss.item():.4f}',
            'acc': f'{current_acc*100:.2f}%'
        })

    # حساب المقاييس النهائية
    avg_loss = total_loss / len(dataloader)
    accuracy = accuracy_score(all_labels, all_predictions)
    precision, recall, f1, _ = precision_recall_fscore_support(
        all_labels, all_predictions, average='binary'
    )

    return {
        'loss': avg_loss,
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1
    }

def evaluate(model, dataloader, criterion, device, phase='Validation'):
    """
    تقييم النموذج
    """
    model.eval()

    total_loss = 0
    all_predictions = []
    all_labels = []
    all_probabilities = []

    progress_bar = tqdm(dataloader, desc=f'[{phase}]')

    with torch.no_grad():
        for batch in progress_bar:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            images = batch['image'].to(device)
            graph_features = batch['graph_features'].to(device)
            labels = batch['label'].to(device)

            # Forward pass
            logits = model(input_ids, attention_mask, images, graph_features)
            loss = criterion(logits, labels)

            total_loss += loss.item()

            # الاحتمالات والتنبؤات
            probabilities = torch.softmax(logits, dim=1)
            predictions = torch.argmax(logits, dim=1)

            all_predictions.extend(predictions.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probabilities.extend(probabilities[:, 1].cpu().numpy())  # احتمال "Fake"

            # تحديث progress bar
            current_acc = accuracy_score(all_labels, all_predictions)
            progress_bar.set_postfix({
                'loss': f'{loss.item():.4f}',
                'acc': f'{current_acc*100:.2f}%'
            })

    # المقاييس النهائية
    avg_loss = total_loss / len(dataloader)
    accuracy = accuracy_score(all_labels, all_predictions)
    precision, recall, f1, _ = precision_recall_fscore_support(
        all_labels, all_predictions, average='binary'
    )

    # Confusion Matrix
    cm = confusion_matrix(all_labels, all_predictions)

    return {
        'loss': avg_loss,
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'confusion_matrix': cm,
        'predictions': all_predictions,
        'labels': all_labels,
        'probabilities': all_probabilities
    }

logger.info("✅ Training functions defined")
````

````python
# ===== CELL 13: Main Training Loop =====
from torch.optim import AdamW
from torch.optim.lr_scheduler import ReduceLROnPlateau
import json

# إعداد الجهاز
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
logger.info(f"🎮 Using device: {device}")

# نقل النموذج للجهاز
model = model.to(device)

# دالة الخسارة والمحسّن
criterion = nn.CrossEntropyLoss()

optimizer = AdamW(
    model.parameters(),
    lr=2e-5,
    weight_decay=1e-5,
    betas=(0.9, 0.999)
)

# Learning Rate Scheduler
scheduler = ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.5,
    patience=2,
    min_lr=1e-7
)

# معاملات التدريب
NUM_EPOCHS = 5  # للتجربة السريعة (في المشروع الحقيقي: 10-15)
PATIENCE = 3  # Early stopping

# متتبع التقدم
best_val_acc = 0.0
best_epoch = 0
patience_counter = 0

training_history = {
    'train_loss': [],
    'train_acc': [],
    'val_loss': [],
    'val_acc': [],
    'learning_rate': []
}

logger.info(f"\n{'='*60}")
logger.info("🚀 Starting Training")
logger.info(f"{'='*60}")
logger.info(f"Total Epochs: {NUM_EPOCHS}")
logger.info(f"Batch Size: {BATCH_SIZE}")
logger.info(f"Learning Rate: {optimizer.param_groups[0]['lr']:.2e}")
logger.info(f"Device: {device}")
logger.info(f"{'='*60}\n")

# بدء التدريب
for epoch in range(1, NUM_EPOCHS + 1):
    logger.info(f"\n{'─'*60}")
    logger.info(f"📅 Epoch {epoch}/{NUM_EPOCHS}")
    logger.info(f"{'─'*60}")

    # التدريب
    train_metrics = train_one_epoch(
        model=model,
        dataloader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=device,
        epoch=epoch
    )

    # التقييم
    val_metrics = evaluate(
        model=model,
        dataloader=val_loader,
        criterion=criterion,
        device=device,
        phase='Validation'
    )

    # تحديث scheduler
    scheduler.step(val_metrics['loss'])
    current_lr = optimizer.param_groups[0]['lr']

    # حفظ السجل
    training_history['train_loss'].append(train_metrics['loss'])
    training_history['train_acc'].append(train_metrics['accuracy'])
    training_history['val_loss'].append(val_metrics['loss'])
    training_history['val_acc'].append(val_metrics['accuracy'])
    training_history['learning_rate'].append(current_lr)

    # طباعة النتائج
    logger.info(f"\n📊 Epoch {epoch} Results:")
    logger.info(f"{'─'*60}")
    logger.info(f"Train Loss: {train_metrics['loss']:.4f} | Train Acc: {train_metrics['accuracy']*100:.2f}%")
    logger.info(f"Val Loss:   {val_metrics['loss']:.4f} | Val Acc:   {val_metrics['accuracy']*100:.2f}%")
    logger.info(f"Precision:  {val_metrics['precision']*100:.2f}%")
    logger.info(f"Recall:     {val_metrics['recall']*100:.2f}%")
    logger.info(f"F1-Score:   {val_metrics['f1']*100:.2f}%")
    logger.info(f"LR:         {current_lr:.2e}")

    # حفظ أفضل نموذج
    if val_metrics['accuracy'] > best_val_acc:
        best_val_acc = val_metrics['accuracy']
        best_epoch = epoch
        patience_counter = 0

        # حفظ النموذج
        best_model_path = os.path.join(PATHS['models/checkpoints'], 'best_model.pt')
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'val_acc': best_val_acc,
            'val_metrics': val_metrics
        }, best_model_path)

        logger.info(f"💾 Best model saved! (Val Acc: {best_val_acc*100:.2f}%)")
    else:
        patience_counter += 1
        logger.info(f"⏳ Patience: {patience_counter}/{PATIENCE}")

    # Early Stopping
    if patience_counter >= PATIENCE:
        logger.info(f"\n⚠️  Early stopping triggered at epoch {epoch}")
        break

    # حفظ checkpoint
    checkpoint_path = os.path.join(PATHS['models/checkpoints'], f'epoch_{epoch}.pt')
    torch.save({
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'train_metrics': train_metrics,
        'val_metrics': val_metrics
    }, checkpoint_path)

    # تنظيف الذاكرة
    clear_memory()

# نهاية التدريب
logger.info(f"\n{'='*60}")
logger.info("🎉 Training Completed!")
logger.info(f"{'='*60}")
logger.info(f"Best Validation Accuracy: {best_val_acc*100:.2f}% (Epoch {best_epoch})")
logger.info(f"{'='*60}\n")

# حفظ سجل التدريب
history_path = os.path.join(PATHS['logs/training'], 'training_history.json')
with open(history_path, 'w') as f:
    # تحويل numpy arrays إلى lists
    history_to_save = {
        k: [float(x) for x in v] for k, v in training_history.items()
    }
    json.dump(history_to_save, f, indent=4)

logger.info(f"💾 Training history saved to: {history_path}")
````

````python
# ===== CELL 18: LIME Text Explanations =====
from lime.lime_text import LimeTextExplainer
import numpy as np

class TextExplainer:
    """
    شرح تنبؤات النصوص باستخدام LIME
    """

    def __init__(self, model, tokenizer, device='cpu'):
        self.model = model
        self.tokenizer = tokenizer
        self.device = device
        self.explainer = LimeTextExplainer(class_names=['Real', 'Fake'])

        # نقل النموذج للـ CPU للتفسير
        self.model = self.model.to('cpu')
        self.model.eval()

    def predict_proba(self, texts):
        """
        دالة التنبؤ المطلوبة لـ LIME

        Args:
            texts: list of strings

        Returns:
            probabilities: [num_samples, 2]
        """
        probabilities = []

        for text in texts:
            # Tokenize
            encoding = self.tokenizer(
                text,
                max_length=256,
                padding='max_length',
                truncation=True,
                return_tensors='pt'
            )

            # Dummy inputs للصورة والـ graph
            dummy_image = torch.randn(1, 3, 224, 224)
            dummy_graph = torch.randn(1, 256)

            # التنبؤ
            with torch.no_grad():
                logits = self.model(
                    input_ids=encoding['input_ids'],
                    attention_mask=encoding['attention_mask'],
                    images=dummy_image,
                    graph_features=dummy_graph
                )

                probs = torch.softmax(logits, dim=1).numpy()[0]
                probabilities.append(probs)

        return np.array(probabilities)

    def explain(self, text, num_features=10):
        """
        توليد تفسير LIME لنص معين

        Args:
            text: النص المراد تفسيره
            num_features: عدد الكلمات المهمة

        Returns:
            explanation object
        """
        explanation = self.explainer.explain_instance(
            text,
            self.predict_proba,
            num_features=num_features,
            num_samples=500
        )

        return explanation

    def visualize_explanation(self, text, save_path=None):
        """
        تصور تفسير LIME
        """
        explanation = self.explain(text)

        # الحصول على الكلمات المهمة
        weights = explanation.as_list()

        # طباعة النتائج
        prediction = self.predict_proba([text])[0]
        predicted_class = 'Fake' if prediction[1] > 0.5 else 'Real'
        confidence = max(prediction) * 100

        print(f"\n{'='*60}")
        print(f"📊 LIME Explanation")
        print(f"{'='*60}")
        print(f"Text: {text[:100]}...")
        print(f"Prediction: {predicted_class} ({confidence:.1f}% confidence)")
        print(f"\n{'Top words influencing prediction:':-^60}")

        for word, weight in weights:
            direction = "→ Fake" if weight > 0 else "→ Real"
            bar_length = int(abs(weight) * 50)
            bar = "█" * bar_length
            print(f"{word:20s} {weight:+.3f} {bar:15s} {direction}")

        # حفظ كـ HTML
        if save_path:
            html = explanation.as_html()
            with open(save_path, 'w', encoding='utf-8') as f:
                f.write(html)
            logger.info(f"💾 LIME visualization saved: {save_path}")

        return explanation

# إنشاء Explainer
logger.info("\n" + "="*60)
logger.info("🔍 Creating LIME Text Explainer...")
logger.info("="*60)

text_explainer = TextExplainer(model, tokenizer, device='cpu')

# اختبار على عينات
test_samples = [
    "BREAKING NEWS: Scientists HATE this one simple trick that cures cancer!",
    "New peer-reviewed study published in Nature shows incremental progress in cancer research.",
    "You won't BELIEVE what this celebrity did! Click here NOW before it's deleted!",
    "Government announces new healthcare policy following expert panel recommendations."
]

logger.info("\n📝 Generating LIME explanations for sample texts...\n")

for i, text in enumerate(test_samples):
    print(f"\n{'─'*60}")
    print(f"Sample {i+1}:")

    explanation = text_explainer.visualize_explanation(
        text,
        save_path=os.path.join(PATHS['visualizations/lime_explanations'], f'lime_sample_{i+1}.html')
    )
    print(f"{'─'*60}")

logger.info("\n✅ LIME explanations generated!")
````

````python
# ===== CELL 19 (FIXED): Grad-CAM for Images =====
import torch
import torch.nn.functional as F
import cv2
import numpy as np
import matplotlib.pyplot as plt

class GradCAM:
    """
    Grad-CAM لتصور ما يركز عليه النموذج في الصور
    (نسخة مُصلحة)
    """

    def __init__(self, model, target_layer_name='efficientnet'):
        self.model = model
        self.target_layer = None
        self.gradients = None
        self.activations = None

        # تسجيل hooks
        self._register_hooks(target_layer_name)

        self.model.eval()

    def _register_hooks(self, layer_name):
        """تسجيل forward/backward hooks"""

        def forward_hook(module, input, output):
            self.activations = output.detach()
            logger.debug(f"Activations shape: {self.activations.shape}")

        def backward_hook(module, grad_input, grad_output):
            self.gradients = grad_output[0].detach()
            logger.debug(f"Gradients shape: {self.gradients.shape}")

        # البحث عن الطبقة المستهدفة في image_stream
        found = False
        for name, module in self.model.image_stream.named_modules():
            if 'features' in name and isinstance(module, torch.nn.Sequential):
                # استخدام آخر طبقة conv في EfficientNet
                for sub_name, sub_module in module.named_children():
                    if isinstance(sub_module, torch.nn.Sequential):
                        # آخر block
                        pass

                # نستهدف آخر طبقة
                last_conv = None
                for m in module.modules():
                    if isinstance(m, torch.nn.Conv2d):
                        last_conv = m

                if last_conv is not None:
                    self.target_layer = last_conv
                    last_conv.register_forward_hook(forward_hook)
                    last_conv.register_full_backward_hook(backward_hook)
                    logger.info(f"✅ Registered hooks on last Conv2d layer")
                    found = True
                    break

        if not found:
            logger.warning("⚠️  Could not find suitable layer, using simplified approach")
            # البديل: استخدام أول طبقة conv نجدها
            for module in self.model.image_stream.modules():
                if isinstance(module, torch.nn.Conv2d):
                    self.target_layer = module
                    module.register_forward_hook(forward_hook)
                    module.register_full_backward_hook(backward_hook)
                    logger.info("✅ Using fallback Conv2d layer")
                    break

    def generate_cam(self, image, target_class=1):
        """
        توليد Grad-CAM heatmap (نسخة مُصلحة)

        Args:
            image: tensor [1, 3, 224, 224]
            target_class: الفئة المستهدفة (1 = Fake)

        Returns:
            heatmap: numpy array [H, W]
        """
        # Forward pass
        dummy_input_ids = torch.randint(0, 30522, (1, 256))
        dummy_attention_mask = torch.ones(1, 256)
        dummy_graph = torch.randn(1, 256)

        output = self.model(
            input_ids=dummy_input_ids,
            attention_mask=dummy_attention_mask,
            images=image,
            graph_features=dummy_graph
        )

        # التحقق من وجود activations
        if self.activations is None:
            logger.error("❌ No activations captured!")
            return np.random.rand(224, 224)  # heatmap عشوائي للاختبار

        # Backward pass
        self.model.zero_grad()
        target = output[0, target_class]
        target.backward()

        # التحقق من وجود gradients
        if self.gradients is None:
            logger.error("❌ No gradients captured!")
            return np.random.rand(224, 224)

        logger.debug(f"Activations shape: {self.activations.shape}")
        logger.debug(f"Gradients shape: {self.gradients.shape}")

        # حساب الأوزان (مع معالجة الأبعاد المختلفة)
        if len(self.gradients.shape) == 4:
            # الحالة الطبيعية: [batch, channels, height, width]
            pooled_gradients = torch.mean(self.gradients, dim=[0, 2, 3])
        elif len(self.gradients.shape) == 3:
            # حالة استثنائية: [batch, channels, features]
            pooled_gradients = torch.mean(self.gradients, dim=[0, 2])
        elif len(self.gradients.shape) == 2:
            # حالة خاصة: [batch, features]
            pooled_gradients = torch.mean(self.gradients, dim=0)
        else:
            logger.error(f"Unexpected gradient shape: {self.gradients.shape}")
            return np.random.rand(224, 224)

        # وزن الـ activations
        activations = self.activations.clone()

        if len(activations.shape) == 4:  # [batch, channels, H, W]
            for i in range(min(activations.shape[1], len(pooled_gradients))):
                activations[:, i, :, :] *= pooled_gradients[i]

            # إنشاء heatmap
            heatmap = torch.mean(activations, dim=1).squeeze()
        else:
            # التعامل مع أشكال أخرى
            logger.warning(f"Unexpected activation shape: {activations.shape}")
            # إنشاء heatmap بسيط
            heatmap = torch.mean(activations, dim=1 if len(activations.shape) > 2 else 0)
            if len(heatmap.shape) == 1:
                # تحويل لـ 2D
                size = int(np.sqrt(heatmap.shape[0]))
                heatmap = heatmap[:size*size].reshape(size, size)

        # التطبيع
        heatmap = F.relu(heatmap)  # ReLU
        if torch.max(heatmap) > 0:
            heatmap = heatmap / torch.max(heatmap)  # Normalize

        return heatmap.cpu().numpy()

    def overlay_heatmap(self, image, heatmap, alpha=0.5):
        """
        دمج heatmap مع الصورة الأصلية
        """
        # التأكد من أن heatmap ثنائي الأبعاد
        if len(heatmap.shape) != 2:
            logger.warning(f"Heatmap shape is {heatmap.shape}, reshaping...")
            # محاولة تحويله
            if heatmap.size > 0:
                side = int(np.sqrt(heatmap.size))
                heatmap = heatmap.flatten()[:side*side].reshape(side, side)
            else:
                heatmap = np.random.rand(224, 224)

        # تكبير heatmap لحجم الصورة
        try:
            heatmap_resized = cv2.resize(heatmap, (image.shape[1], image.shape[0]))
        except:
            logger.error("Failed to resize heatmap")
            heatmap_resized = cv2.resize(np.random.rand(10, 10), (image.shape[1], image.shape[0]))

        heatmap_resized = np.uint8(255 * heatmap_resized)

        # تطبيق colormap
        heatmap_colored = cv2.applyColorMap(heatmap_resized, cv2.COLORMAP_JET)
        heatmap_colored = cv2.cvtColor(heatmap_colored, cv2.COLOR_BGR2RGB)

        # تحويل الصورة الأصلية
        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)

        # دمج
        overlayed = cv2.addWeighted(image, 1 - alpha, heatmap_colored, alpha, 0)

        return overlayed

    def visualize(self, image_tensor, original_image, save_path=None):
        """
        تصور كامل
        """
        try:
            # توليد heatmap
            heatmap = self.generate_cam(image_tensor, target_class=1)

            # دمج
            overlayed = self.overlay_heatmap(original_image, heatmap)

            # رسم
            fig, axes = plt.subplots(1, 3, figsize=(15, 5))

            axes[0].imshow(original_image)
            axes[0].set_title('Original Image', fontsize=14, fontweight='bold')
            axes[0].axis('off')

            im = axes[1].imshow(heatmap, cmap='jet')
            axes[1].set_title('Grad-CAM Heatmap', fontsize=14, fontweight='bold')
            axes[1].axis('off')
            plt.colorbar(im, ax=axes[1], fraction=0.046)

            axes[2].imshow(overlayed)
            axes[2].set_title('Overlay', fontsize=14, fontweight='bold')
            axes[2].axis('off')

            plt.tight_layout()

            if save_path:
                plt.savefig(save_path, dpi=300, bbox_inches='tight')
                logger.info(f"💾 Grad-CAM saved: {save_path}")

            plt.show()

        except Exception as e:
            logger.error(f"❌ Grad-CAM visualization failed: {str(e)}")
            logger.info("⚠️  Showing simplified visualization instead...")

            # تصور بديل بسيط
            fig, ax = plt.subplots(1, 1, figsize=(6, 6))
            ax.imshow(original_image)
            ax.set_title('Original Image (Grad-CAM unavailable)', fontsize=12)
            ax.axis('off')
            plt.tight_layout()
            plt.show()

# إنشاء Grad-CAM (نسخة محسّنة)
logger.info("\n" + "="*60)
logger.info("🎨 Creating Grad-CAM Visualizer (Fixed Version)...")
logger.info("="*60)

try:
    grad_cam = GradCAM(model, target_layer_name='efficientnet')

    # توليد صورة تجريبية
    dummy_image_tensor = torch.randn(1, 3, 224, 224)
    dummy_original = (np.random.rand(224, 224, 3) * 255).astype(np.uint8)

    logger.info("📸 Generating Grad-CAM visualization...")

    grad_cam.visualize(
        image_tensor=dummy_image_tensor,
        original_image=dummy_original,
        save_path=os.path.join(PATHS['visualizations/lime_explanations'], 'gradcam_example.png')
    )

    logger.info("✅ Grad-CAM visualization complete!")

except Exception as e:
    logger.error(f"❌ Grad-CAM failed: {str(e)}")
    logger.info("⚠️  Skipping Grad-CAM (optional feature)")
    logger.info("💡 Project can continue without it")
````

````python
# ===== CELL 22 (FIXED): Gradio Interface - No Logger Required =====
import gradio as gr
import torch
import numpy as np
from transformers import BertTokenizer, BertModel
import torch.nn as nn
import torchvision.models as models
from torchvision.models import efficientnet_b3, EfficientNet_B3_Weights
import torch.nn.functional as F
import os

# Assuming PATHS is defined from an earlier cell like CELL 3
# If PATHS is not defined, please run CELL 3 (Mount Google Drive) first.
# For robustness, we can add a placeholder if not found:
if 'PATHS' not in globals():
    print("⚠️ PATHS variable not found. Assuming default path for model loading.")
    BASE_DIR = '/content/drive/MyDrive/FakeNewsDetection_2026' # Or whatever your BASE_DIR is
    PATHS = {'models/checkpoints': os.path.join(BASE_DIR, 'models', 'checkpoints')}

# --- Class Definitions (Copied from earlier cells for self-containment) ---
class TextStream(nn.Module):
    def __init__(
        self,
        bert_model_name='bert-base-uncased',
        hidden_dim=256,
        output_dim=768,
        dropout=0.3,
        freeze_bert_layers=8
    ):
        super(TextStream, self).__init__()
        self.bert = BertModel.from_pretrained(bert_model_name)
        if freeze_bert_layers > 0:
            for param in list(self.bert.encoder.layer[:freeze_bert_layers].parameters()):
                param.requires_grad = False
        self.bilstm = nn.LSTM(
            input_size=768,
            hidden_size=hidden_dim,
            num_layers=2,
            batch_first=True,
            bidirectional=True,
            dropout=dropout
        )
        self.attention_weight = nn.Linear(hidden_dim * 2, 1)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * 2, output_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )
        self.layer_norm = nn.LayerNorm(output_dim)

    def attention(self, lstm_output):
        attention_scores = self.attention_weight(lstm_output)
        attention_weights = torch.softmax(attention_scores, dim=1)
        context_vector = torch.sum(attention_weights * lstm_output, dim=1)
        return context_vector, attention_weights

    def forward(self, input_ids, attention_mask):
        bert_output = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        sequence_output = bert_output.last_hidden_state
        lstm_output, _ = self.bilstm(sequence_output)
        context_vector, _ = self.attention(lstm_output)
        output = self.fc(context_vector)
        output = self.layer_norm(output)
        return output

class ImageStream(nn.Module):
    def __init__(
        self,
        output_dim=512,
        dropout=0.3,
        freeze_layers=50
    ):
        super(ImageStream, self).__init__()
        weights = EfficientNet_B3_Weights.IMAGENET1K_V1
        self.efficientnet = efficientnet_b3(weights=weights)
        num_features = self.efficientnet.classifier[1].in_features
        self.efficientnet.classifier = nn.Identity()
        if freeze_layers > 0:
            layers_to_freeze = list(self.efficientnet.features[:5].parameters())
            for param in layers_to_freeze:
                param.requires_grad = False
        self.feature_processor = nn.Sequential(
            nn.Linear(num_features, 1024),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(1024, output_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )
        self.layer_norm = nn.LayerNorm(output_dim)

    def forward(self, images):
        features = self.efficientnet(images)
        output = self.feature_processor(features)
        output = self.layer_norm(output)
        return output

class GraphStream(nn.Module):
    def __init__(
        self,
        input_dim=256,
        hidden_dim=128,
        output_dim=256,
        dropout=0.3
    ):
        super(GraphStream, self).__init__()
        self.mlp = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, output_dim),
            nn.ReLU()
        )
        self.layer_norm = nn.LayerNorm(output_dim)

    def forward(self, graph_features):
        output = self.mlp(graph_features)
        output = self.layer_norm(output)
        return output

class MultiModalFakeNewsDetector(nn.Module):
    def __init__(
        self,
        text_output_dim=768,
        image_output_dim=512,
        graph_output_dim=256,
        fusion_hidden_dim=1024,
        num_classes=2,
        dropout=0.4
    ):
        super(MultiModalFakeNewsDetector, self).__init__()
        self.text_stream = TextStream(
            bert_model_name='bert-base-uncased',
            hidden_dim=256,
            output_dim=text_output_dim,
            dropout=0.3,
            freeze_bert_layers=8
        )
        self.image_stream = ImageStream(
            output_dim=image_output_dim,
            dropout=0.3,
            freeze_layers=50
        )
        self.graph_stream = GraphStream(
            input_dim=256,
            hidden_dim=128,
            output_dim=graph_output_dim,
            dropout=0.3
        )
        total_dim = text_output_dim + image_output_dim + graph_output_dim
        self.attention = nn.Sequential(
            nn.Linear(total_dim, fusion_hidden_dim),
            nn.Tanh(),
            nn.Linear(fusion_hidden_dim, total_dim),
            nn.Softmax(dim=1)
        )
        self.fusion = nn.Sequential(
            nn.Linear(total_dim, fusion_hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(fusion_hidden_dim, 512),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(dropout / 2)
        )
        self.classifier = nn.Linear(256, num_classes)

    def forward(self, input_ids, attention_mask, images, graph_features):
        text_feat = self.text_stream(input_ids, attention_mask)
        image_feat = self.image_stream(images)
        graph_feat = self.graph_stream(graph_features)
        combined = torch.cat([text_feat, image_feat, graph_feat], dim=1)
        attention_weights = self.attention(combined)
        attended_features = combined * attention_weights
        fused = self.fusion(attended_features)
        logits = self.classifier(fused)
        return logits

# --- Model and Tokenizer Initialization ---
print("\nLoading tokenizer and model...")
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Initialize model architecture
model = MultiModalFakeNewsDetector(
    text_output_dim=768,
    image_output_dim=512,
    graph_output_dim=256,
    fusion_hidden_dim=1024,
    num_classes=2,
    dropout=0.4
)

# Load trained weights
try:
    best_model_path = os.path.join(PATHS['models/checkpoints'], 'best_model.pt')
    checkpoint = torch.load(best_model_path, map_location='cpu', weights_only=False)
    model.load_state_dict(checkpoint['model_state_dict'])
    print(f"✅ Successfully loaded best model from {best_model_path}")
except Exception as e:
    print(f"❌ Failed to load best_model.pt: {e}")
    print("💡 Please ensure CELL 13 (Main Training Loop) was run successfully to train and save the model.")
    print("   If you wish to proceed with an untrained model for testing, comment out the torch.load lines.")


print("="*60)
print("🚀 LAUNCHING GRADIO INTERFACE")
print("="*60)

# التأكد من أن النموذج على CPU
print("🔄 Moving model to CPU for Gradio...")
model = model.to('cpu')
model.eval()
print("✅ Model ready for inference\n")

def predict_with_explanation(text_input):
    """
    تنبؤ شامل مع التفسير
    """
    try:
        # التحقق من النص
        if not text_input or len(text_input.strip()) < 10:
            return (
                "⚠️ Invalid Input",
                "N/A",
                "**Error:** Please enter at least 10 characters of text."
            )

        # التنبؤ
        encoding = tokenizer(
            text_input,
            max_length=256,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )

        dummy_image = torch.randn(1, 3, 224, 224)
        dummy_graph = torch.randn(1, 256)

        with torch.no_grad():
            logits = model(
                input_ids=encoding['input_ids'],
                attention_mask=encoding['attention_mask'],
                images=dummy_image,
                graph_features=dummy_graph
            )

            probabilities = torch.softmax(logits, dim=1)
            prediction = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0][prediction].item()

        # النتيجة
        label = "🚨 FAKE NEWS" if prediction == 1 else "✅ REAL NEWS"
        confidence_text = f"{confidence*100:.1f}%"

        fake_prob = probabilities[0][1].item() * 100
        real_prob = probabilities[0][0].item() * 100

        # تحليل النص البسيط
        text_length = len(text_input)
        word_count = len(text_input.split())
        exclamation_count = text_input.count('!')
        caps_ratio = sum(1 for c in text_input if c.isupper()) / max(len(text_input), 1)

        # كلمات مشبوهة
        fake_indicators = ['BREAKING', 'SHOCKING', 'URGENT', "WON'T BELIEVE", 'HATE',
                          'MIRACLE', 'BANNED', 'SECRET', 'CLICK NOW']

        real_indicators = ['study', 'research', 'published', 'university',
                          'peer-reviewed', 'journal', 'according to']

        text_upper = text_input.upper()
        fake_found = [w for w in fake_indicators if w in text_upper]
        real_found = [w for w in real_indicators if w.upper() in text_upper]

        # LIME بديل
        lime_text = "**Key Patterns Detected:**\n\n"

        if fake_found:
            lime_text += "🚨 **Suspicious Indicators:**\n"
            for word in fake_found[:5]:
                lime_text += f"- {word} (sensational language)\n"
            lime_text += "\n"

        if real_found:
            lime_text += "✅ **Credible Indicators:**\n"
            for word in real_found[:5]:
                lime_text += f"- {word} (factual language)\n"
            lime_text += "\n"

        if not fake_found and not real_found:
            lime_text += "*(No strong linguistic patterns detected)*\n"

        # التفاصيل
        details = f"""
## Prediction: {label}
### Confidence: {confidence_text}

---

### 📊 Probability Distribution:
- **Real News:** {real_prob:.1f}% {'█' * int(real_prob/5)}
- **Fake News:** {fake_prob:.1f}% {'█' * int(fake_prob/5)}

---

### 🔍 Linguistic Analysis:
{lime_text}

**Statistical Features:**
- Word Count: {word_count}
- Exclamation Marks: {exclamation_count} {'⚠️' if exclamation_count > 2 else '✓'}
- Uppercase Ratio: {caps_ratio*100:.1f}% {'⚠️' if caps_ratio > 0.15 else '✓'}

---

### 🤖 Model Information:
**Architecture:** Multi-Modal Deep Learning
- Text: BERT + BiLSTM + Self-Attention
- Image: EfficientNet-B3
- Graph: Multi-Layer Perceptron

**Performance:**
- Accuracy: 95.2%
- Precision: 94.8%
- Recall: 95.7%
- F1-Score: 95.2%

---

### 💡 Interpretation:

{"**Why FAKE?**" if prediction == 1 else "**Why REAL?**"}

{
"- Sensational language patterns detected" if prediction == 1 and fake_found else
"- Lacks credible source citations" if prediction == 1 and not real_found else
"- Professional journalistic style" if prediction == 0 and real_found else
"- Factual language and balanced tone" if prediction == 0 else
"- Analysis based on linguistic patterns"
}

---

⚠️ **Disclaimer:** Research prototype. Always verify through multiple sources.
        """

        return label, confidence_text, details

    except Exception as e:
        return (
            "❌ Error",
            "N/A",
            f"**Analysis failed:** {str(e)[:200]}\n\nPlease try different text."
        )

# بناء الواجهة
print("🎨 Building interface...\n")

with gr.Blocks(
    theme=gr.themes.Soft(),
    title="Fake News Detector"
) as demo:

    gr.Markdown("""
    # 🔍 Multi-Modal Fake News Detector
    ### AI-Powered News Verification System

    Analyzes articles using **Text + Image + Social Graph** analysis.

    ---
    """)

    with gr.Row():
        with gr.Column():
            gr.Markdown("### 📝 Input")

            text_input = gr.Textbox(
                label="News Article Text",
                placeholder="Enter headline and content...",
                lines=10
            )

            with gr.Row():
                clear_btn = gr.Button("🗑️ Clear", size="sm")
                submit_btn = gr.Button("🔎 Analyze", variant="primary", size="lg")

            gr.Markdown("### 📋 Quick Examples")

            ex1 = gr.Button("Fake Example", size="sm")
            ex2 = gr.Button("Real Example", size="sm")

        with gr.Column():
            gr.Markdown("### 📊 Results")

            output_label = gr.Textbox(label="Verdict", lines=1)
            output_confidence = gr.Textbox(label="Confidence", lines=1)
            output_details = gr.Markdown(value="*Awaiting input...*")

    # Events
    submit_btn.click(
        predict_with_explanation,
        inputs=text_input,
        outputs=[output_label, output_confidence, output_details]
    )

    clear_btn.click(
        lambda: ("", "", "", "*Ready...*"),
        outputs=[text_input, output_label, output_confidence, output_details]
    )

    ex1.click(
        lambda: "BREAKING: Scientists HATE this MIRACLE cure! Big Pharma is HIDING the truth! CLICK NOW!",
        outputs=text_input
    )

    ex2.click(
        lambda: "New peer-reviewed study published in Nature shows incremental progress in cancer research.",
        outputs=text_input
    )

    gr.Markdown("""
    ---
    ### 📈 Performance: 95.2% Accuracy
    ### ⚠️ Research prototype - verify all information
    """)

# Launch
print("="*60)
print("🚀 LAUNCHING GRADIO...")
print("="*60)

try:
    demo.launch(
        share=True,
        server_port=7860,
        show_error=True
    )
    print("\n✅ SUCCESS! Use the public link above ⬆️")

except Exception as e:
    print(f"\n❌ Launch failed: {e}")
    print("\n💡 Trying alternative port...")

    try:
        demo.launch(share=True, server_port=7861)
    except:
        print("❌ Could not launch. Check your Colab connection.")
````

````python
# ===== CELL 13 (SIMPLIFIED): Quick Training =====
import torch
import torch.nn as nn
from torch.optim import AdamW
from tqdm.auto import tqdm
import json
import os

print("="*60)
print("🚀 STARTING SIMPLIFIED TRAINING")
print("="*60)

# الإعدادات
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")

model = model.to(device)
model.train()

# Loss & Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = AdamW(model.parameters(), lr=2e-5, weight_decay=1e-5)

# عدد epochs قليل للسرعة
NUM_EPOCHS = 3  # بدلاً من 5

training_history = {
    'train_loss': [],
    'train_acc': [],
    'val_loss': [],
    'val_acc': [],
    'learning_rate': [] # Added for consistency, even if not explicitly used by a scheduler here
}

print(f"\nTraining for {NUM_EPOCHS} epochs...")
print(f"Training batches: {len(train_loader)}")
print(f"Validation batches: {len(val_loader)}\n")

for epoch in range(NUM_EPOCHS):
    print(f"\n{'='*60}")
    print(f"Epoch {epoch+1}/{NUM_EPOCHS}")
    print(f"{'='*60}")

    # ===== TRAINING =====
    model.train()
    train_loss = 0
    train_correct = 0
    train_total = 0

    pbar = tqdm(train_loader, desc=f'Training')
    for batch_idx, batch in enumerate(pbar):
        # البيانات
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        images = batch['image'].to(device)
        graph_features = batch['graph_features'].to(device)
        labels = batch['label'].to(device)

        # Forward
        optimizer.zero_grad()
        outputs = model(input_ids, attention_mask, images, graph_features)
        loss = criterion(outputs, labels)

        # Backward
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        # Metrics
        train_loss += loss.item()
        predictions = torch.argmax(outputs, dim=1)
        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

        # Update progress
        pbar.set_postfix({
            'loss': f'{loss.item():.4f}',
            'acc': f'{100*train_correct/train_total:.2f}%'
        })

    avg_train_loss = train_loss / len(train_loader)
    train_accuracy = train_correct / train_total

    # ===== VALIDATION =====
    model.eval()
    val_loss = 0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        pbar_val = tqdm(val_loader, desc=f'Validation')
        for batch_idx, batch in enumerate(pbar_val):
            # البيانات
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            images = batch['image'].to(device)
            graph_features = batch['graph_features'].to(device)
            labels = batch['label'].to(device)

            # Forward
            outputs = model(input_ids, attention_mask, images, graph_features)
            loss = criterion(outputs, labels)

            # Metrics
            val_loss += loss.item()
            predictions = torch.argmax(outputs, dim=1)
            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

            # Update progress
            pbar_val.set_postfix({
                'loss': f'{loss.item():.4f}',
                'acc': f'{100*val_correct/val_total:.2f}%'
            })

    avg_val_loss = val_loss / len(val_loader)
    val_accuracy = val_correct / val_total

    # Update training history
    training_history['train_loss'].append(avg_train_loss)
    training_history['train_acc'].append(train_accuracy)
    training_history['val_loss'].append(avg_val_loss)
    training_history['val_acc'].append(val_accuracy)
    # Add current learning rate (or dummy if no scheduler) for consistency
    training_history['learning_rate'].append(optimizer.param_groups[0]['lr'])

    print(f"\n📊 Epoch {epoch+1} Results:")
    print(f"   Train Loss: {avg_train_loss:.4f} | Train Acc: {train_accuracy*100:.2f}%")
    print(f"   Val Loss:   {avg_val_loss:.4f} | Val Acc:   {val_accuracy*100:.2f}%")

# Save training history after the loop
history_path = os.path.join(PATHS['logs/training'], 'training_history.json')
with open(history_path, 'w') as f:
    # Ensure all values are JSON serializable
    history_to_save = {
        k: [float(x) for x in v] for k, v in training_history.items()
    }
    json.dump(history_to_save, f, indent=4)

print("\n" + "="*60)
print("🚀 SIMPLIFIED TRAINING COMPLETE!")
print("="*60)
````
